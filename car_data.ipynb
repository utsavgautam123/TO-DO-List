```python
import pandas as pd
df=pd.read_csv("Car.csv")
df.head()
```

```python
df.info()
```

```python
df.drop(columns=['source_url', 'battery_type'], inplace=True)
```

```python
df.head()
```

```python
df.head()
```

```python
df['model'] = df['model'].fillna('Unknown')
df.head()
```

```python
df['fast_charging_power_kw_dc'] = df['fast_charging_power_kw_dc'].fillna(df['fast_charging_power_kw_dc'].median())
df.head()
```

```python
df['fast_charge_port'] = df['fast_charge_port'].fillna(df['fast_charge_port'].mode()[0])
df['towing_capacity_kg'] = df['towing_capacity_kg'].fillna(0)

df.head()
```

```python
df.info()
```

```python
df["brand"].nunique()
df['number_of_cells'] = df['number_of_cells'].fillna(0)


df.head()
```

```python
df['cargo_volume_l'] = df['cargo_volume_l'].str.strip()
df.head()
```

```python
df.info()
```

```python
# df['cargo_volume_l'] = df['cargo_volume_l'].str.extract(r'(\d+)')
df['cargo_volume_l'].unique()
```

```python
df.info()
```

```python
df['torque_nm'] = df['torque_nm'].fillna(df['torque_nm'].median())
df.head()
```

```python
df.info()
```
